```python
from langgraph.store.memory import InMemoryStore
```

```python
store=InMemoryStore()
```

```python
namespace=('user','u1')
```

```python
store.put(namespace,'1',{'data':'User like pizza'})
store.put(namespace,'2',{'data':'User like python programming'})
```

```python
namespace2=('user','u2')
```

```python
store.put(namespace2,'1',{'data':'user like pasta'})
store.put(namespace2,'2',{'data':'user prefers grid like style'})
```

```python
# fetch memory

store.get(namespace,'1')
```

```python
# fetch all the memory

item=store.search(namespace)
for items in item:
    print(items.value)
```

```python
item=store.search(namespace2)
for items in item:
    print(items.value)
```
